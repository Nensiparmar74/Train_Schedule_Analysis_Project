# Train Schedule Analysis
This notebook reproduces core analysis from the supplied task list.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
DATA = ROOT / 'data' / 'Dataset1.csv'
df = pd.read_csv(DATA)
print('Shape:', df.shape)
print('Unique train numbers:', df['Train_No'].nunique())
df.head()
```

## Data quality review

```
print('Missing values:\n', df.isna().sum())
print('Exact duplicate rows:', df.duplicated().sum())
print('Dtypes:\n', df.dtypes)
```

## Train-level stop counts and endpoints

```
ordered = df.drop_duplicates().sort_values(['Train_No','Route_Number','SN'])
train_summary = ordered.groupby(['Train_No','Route_Number']).agg(
    Starting_Station=('Station_Name','first'), Ending_Station=('Station_Name','last'),
    Stop_Count=('Station_Name','size')).reset_index()
train_summary['Intermediate_Stops'] = (train_summary['Stop_Count'] - 2).clip(lower=0)
train_summary.sort_values('Stop_Count', ascending=False).head(10)
```

## Station frequency

```
station_frequency = (ordered.groupby(['Station_Code','Station_Name'])['Train_No']
                    .nunique().sort_values(ascending=False).rename('Unique_Trains').reset_index())
station_frequency.head(15)
```
